```python
import pandas as pd
```

```python
df = pd.read_csv("hospital_appointments.csv")
df
```

```python
df.info()
```

## **DATA CLEANING**

#### **1.Fixing Date**

```python
df['DOB'] = pd.to_datetime(df['DOB'] , errors = "coerce" , format = 'mixed')
df
```

```python
df['Appointment_Date'] = pd.to_datetime(df['Appointment_Date'] , errors = 'coerce' , format = 'mixed')
df
```

#### **2.Removing Duplicates**

```python
df.drop_duplicates(inplace=True)
df
```

```python
df['Gender'] = df['Gender'].replace({'M' : 'Male'})
df
```

```python
df.info()
```

#### **3.Fixing Fee Column** 

```python
# changing data type from str to int
```

```python
df["Fee"] = df["Fee"].str.replace("₹", "", regex=False).astype(int)
df
```

```python
df.to_csv("hospital_appointments_cleaned.csv", index=False)
```
