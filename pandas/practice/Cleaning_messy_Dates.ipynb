```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("employee_dataset.csv")
df
```

# **DATA CLEANING**

### **1.String Mapping**

```python
df['Gender'] = df['Gender'].replace({'M' :'Male'})
df
```

### **2.Removing Duplicates**

```python
df.drop_duplicates(inplace = True)
df
```

```python
df.drop_duplicates(subset=["Employee_Name"], inplace=True)
```

```python
df
```

### **3.Fixing Salary**

```python
df['Salary'] = df['Salary'].str.replace('₹' , "" , regex = False).str.replace(',' , '' , regex = False)
df
```

```python
df["Salary"] = df["Salary"].astype(int)
```

### **4.Fixing Dates**

```python
df['Joining_Date'] = pd.to_datetime(df["Joining_Date"], format = 'mixed' , errors = 'coerce')
df
```

# **VISUALIZATION**

### **1.Average Salary By Department**

```python
avg_salary = df.groupby('Department')['Salary'].mean()
avg_salary
```

```python
plt.bar(avg_salary.index , avg_salary.values , color = "#994B68")
plt.xlabel("Department")
plt.ylabel("Salary")
plt.title("Average Salary By Department")
plt.show()
```

### **2.How many employees are there in each department?**

```python
employee_count = df.groupby("Department")["Employee_Name"].count()
employee_count
```

```python
plt.bar(employee_count.index , employee_count.values ,color = "#EDDFDF")
plt.xlabel("Department")
plt.ylabel("No.of employees")
plt.title("How many employees are there in each department")
plt.show()
```

### **INSIGHT**
- HR , IT AND FINANCE HAVE EQUAL NUMBER OF EMPLOYEES WHEREAS SALES HAS THE LEAST NUMBER OF EMPLOYEES.

### **3.How many males and females are there?**

```python
gender_count = df['Gender'].value_counts()
gender_count
```

```python
plt.bar(gender_count.index , gender_count.values ,color = "#EDDFDF")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("How many males and females are there")
plt.show()
```

### **4.Salary Spread (Box Plot)**

```python
plt.boxplot(df["Salary"])

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```

### **5.Salary by Department (Box Plot)**

```python
departments = df["Department"].unique()

salary_data = [
    df[df["Department"] == dept]["Salary"]
    for dept in departments
]

plt.boxplot(salary_data, tick_labels=departments)

plt.title("Salary Distribution by Department")
plt.xlabel("Department")
plt.ylabel("Salary")

plt.show()
```
