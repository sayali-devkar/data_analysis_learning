```python
import pandas as pd
```

```python
df = pd.read_csv("employee_dataset.csv")
df
```

# **DATA CLEANING**

### **1.String Mapping**

```python
df['Gender'] = df['Gender'].replace({'M' :'Male'})
df
```

### **2.Removing Duplicates**

```python
df.drop_duplicates(inplace = True)
df
```

```python
df.drop_duplicates(subset=["Employee_Name"], inplace=True)
```

```python
df
```

### **3.Fixing Salary**

```python
df['Salary'] = df['Salary'].str.replace('₹' , "" , regex = False).str.replace(',' , '' , regex = False)
df
```

### **4.Fixing Dates**

```python
df['Joining_Date'] = pd.to_datetime(df["Joining_Date"], format = 'mixed' , errors = 'coerce')
df
```
