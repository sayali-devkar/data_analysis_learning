```python
import pandas as pd
```

```python
data = {
    'name': ['Aarav', 'Sneha', 'Ranjeet', 'Priya', 'Karan', 'Ankit', 'Divya', 'Rohan'],
    'city': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Nagpur', 'Mumbai', 'Delhi', 'Nagpur'],
    'department': ['HR', 'Tech', 'Finance', 'Tech', 'HR', 'Finance', 'Tech', 'HR'],
    'salary': [30000, 75000, 45000, 68000, 32000, 50000, 80000, 28000]
}
df = pd.DataFrame(data)
df
```

```python
df[df["city"].isin(["Mumbai" , "Pune" , "Nagpur"])]
```

```python
df[df["name"].str.contains("an")]
```

```python
df[~df["department"].isin(["Tech","Finance"])]
```
