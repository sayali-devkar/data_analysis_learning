```python
import pandas as pd

data = {
    'name': ['Aarav', 'Sneha', 'Rahul', 'Priya', 'Karan', 'Mehta', 'Divya', 'Rohan'],
    'age': [22, 25, 19, 30, 21, 28, 24, 35],
    'city': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Mumbai', 'Delhi', 'Pune', 'Mumbai'],
    'salary': [25000, 45000, 20000, 60000, 22000, 50000, 35000, 70000],
    'experience': [1, 3, 0, 6, 1, 4, 2, 8]
}

df = pd.DataFrame(data)
df
```

```python
#1 
df[(df["city"] == "Pune") & (df["salary"] > 30000)]
```

```python
#2
df[df[ 20 <= ["age"] <= 26]]
```

```python
#3
df[(df["experience"] > 2) & (df["city"] != "Delhi")]
```

```python
# Option A — using &
df[(df["age"] >= 20) & (df["age"] <= 26)]
```

```python
# Option B — using .between():
df[df["age"].between(20, 26)]
```

## **Practice Data NO.2**

```python
import pandas as pd

data = {
    'student': ['Riya', 'Arjun', 'Sneha', 'Dev', 'Pooja', 'Karan', 'Meera', 'Sahil'],
    'marks': [85, 42, 91, 55, 78, 33, 88, 60],
    'grade': ['A', 'D', 'A', 'C', 'B', 'F', 'A', 'C'],
    'subject': ['Math', 'Science', 'Math', 'English', 'Science', 'Math', 'English', 'Science'],
    'attendance': [95, 60, 98, 72, 85, 45, 92, 78]
}

df = pd.DataFrame(data)
df
```

```python
#1
df[(df["grade"] == "A") | (df["grade"] == "B")]
```

```python
#2
df[(df["attendance"] < 75 ) & (df["marks"] < 60)]
```

```python
#3
df[((df["subject"] == "Math") | (df["subject"] == "Science")) & (df["marks"] > 50)]
```

### **Practice data no. 3**

```python
import pandas as pd

data = {
    'product': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Earbuds', 'Monitor', 'Keyboard', 'Mouse'],
    'brand': ['Dell', 'Samsung', 'Apple', 'Apple', 'Sony', 'LG', 'Dell', 'Logitech'],
    'price': [75000, 25000, 60000, 35000, 8000, 20000, 3000, 1500],
    'rating': [4.5, 4.2, 4.8, 3.9, 4.1, 4.3, 3.7, 4.6],
    'stock': [10, 0, 5, 8, 0, 3, 15, 20]
}

df = pd.DataFrame(data)
df
```

```python
#1
df[(df["stock"] > 0) & (df["rating"] > 4.2)]
```

```python
#2
df[((df["brand"] == "Apple") | (df["brand"] == "Sony")) & (df["price"] > 20000)]
```

```python
#3
df[(df["brand"] != "Dell") & (df["price"].between(5000 , 40000))]
```
