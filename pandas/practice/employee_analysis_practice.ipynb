```python
import pandas as pd

data = {
    'employee': ['Riya', 'Anand', 'Karan', 'Tanvi', 'Dev', 'Anjali', 'Rohan', 'Sneha', 'Arjun', 'Meera'],
    'department': ['Tech', 'HR', 'Finance', 'Tech', 'HR', 'Finance', 'Tech', 'HR', 'Finance', 'Tech'],
    'city': ['Pune', 'Mumbai', 'Delhi', 'Nagpur', 'Pune', 'Mumbai', 'Delhi', 'Nagpur', 'Pune', 'Mumbai'],
    'salary': [75000, 30000, 45000, 80000, 32000, 50000, 70000, 28000, 55000, 85000],
    'experience': [3, 1, 4, 5, 2, 3, 4, 1, 6, 7],
    'rating': [4.5, 3.2, 4.0, 4.8, 3.5, 4.1, 4.3, 3.0, 4.6, 4.9]
}
df = pd.DataFrame(data)
df
```

## **1.Count employees in each department**

```python
df.groupby("department")["employee"].count()
```

## **2.Average salary by city**

```python
df.groupby("city")["salary"].mean()
```

## **3.Which employee has the highest rating?**

```python
df.loc[df["rating"].idxmax() , "employee"]
```

## **4.Filter employees from Tech or Finance with salary above 60000**

```python
df[df["department"].isin(["Tech","Finance"]) & (df["salary"] > 60000)]
```

## **5.Filter employees with experience between 3 and 6 who are NOT from HR**

```python
df[~df["department"].isin(["HR"]) & (df["experience"].between(3,6))]
```
