```python
import pandas as pd

data = {
    "Name": ["Amit", "Priya", "Rahul", "Sneha", "Karan"],
    "Maths": [85, 92, 78, 88, 95],
    "Science": [90, 85, 80, 91, 89],
    "English": [78, 88, 84, 90, 87]
}

df = pd.DataFrame(data)
df
```

```python
#1
df.head(3)
```

```python
#2
print("average_marks_of_maths",df["Maths"].mean())
```

```python
#3
df["Total"] = df["Maths"] + df["Science"] + df["English"]
df
```

```python
#4
df["Percentage"] = df["Total"] / 3
df
```

```python
#5
df[df["Percentage"] > 85]
```

```python
#6
df.loc[df["Total"].idxmax()]
```

```python
df.sort_values(by="Percentage", ascending=False)
```
