```python
import pandas as pd

data = {
    'student': ['Riya', 'Anand', 'Karan', 'Tanvi', 'Dev', 'Anjali', 'Rohan', 'Sneha'],
    'branch': ['CS', 'IT', 'CS', 'AIML', 'IT', 'AIML', 'CS', 'IT'],
    'city': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Nagpur', 'Mumbai', 'Delhi', 'Nagpur'],
    'cgpa': [8.5, 6.2, 9.1, 7.8, 6.5, 8.9, 7.2, 9.3],
    'backlogs': [0, 2, 0, 1, 3, 0, 1, 0]
}

df = pd.DataFrame(data)
df
```
