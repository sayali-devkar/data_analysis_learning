```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
flights = sns.load_dataset("flights")
```

```python
flights.head()
```

### **Data Cleaning**

```python
flights.info()
```

## **Data Visualization**

### **1.Passenger Trends Across Year**

```python
passenger_Trend = flights.groupby("year")['passengers'].sum()
passenger_Trend
```

```python
plt.plot(passenger_Trend.index , passenger_Trend.values,color = '#A3655A')
plt.title('Passenger Trends Across Year')
plt.xlabel('Year')
plt.ylabel('Total Number Of Passengers')
plt.show()
```

### **INSIGHT**
- Passengers increasing over years

### **2.In which month Most Passengers Travel**

```python
month_analysis = flights.groupby('month')['passengers'].sum()
month_analysis
```

```python
plt.bar(month_analysis.index , month_analysis.values , color = "#B37368")
plt.title('Which Month Has the Highest Passenger Traffic?')
plt.xlabel('Months')
plt.ylabel('Total Passengers')
plt.xticks(rotation = 45 , ha = 'right')
plt.tight_layout()
plt.show()
```

```python
flights['year'].unique()
```

### **3.How are passenger counts distributed across all 144 monthly observations?**

```python
plt.hist(flights["passengers"], bins=10, color="#A3655A" , edgecolor="black")
plt.title("Distribution of Passenger Counts")
plt.xlabel("Passenger Count")
plt.ylabel("Frequency")
plt.show()
```

### **4.Busiest Month of each year**

```python
busiest_month = flights.groupby('year')['passengers'].idxmax()

busiest_month
```

```python
flights.loc[busiest_month]
```

### **Business Insight**

- Passenger demand consistently peaks during July and August, indicating these are the airline's busiest travel months. Additionally, peak passenger counts increase almost every year, suggesting a steady growth in air travel demand.
