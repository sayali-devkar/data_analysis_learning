```python
import pandas as pd
```

```python
data = {
    'product': ['Laptop', 'Phone', 'Tablet', 'Watch', 'Earbuds', 'Monitor', 'Keyboard', 'Mouse', 'Charger', 'Speaker'],
    'category': ['Electronics', 'Electronics', 'Electronics', 'Accessories', 'Accessories', 'Electronics', 'Accessories', 'Accessories', 'Accessories', 'Electronics'],
    'brand': ['Dell', 'Samsung', 'Apple', 'Fossil', 'Sony', 'LG', 'Logitech', 'Logitech', 'Anker', 'Sony'],
    'price': [75000, 25000, 60000, 15000, 8000, 20000, 3000, 1500, 2000, 12000],
    'rating': [4.5, 4.2, 4.8, 3.9, 4.1, 4.3, 3.7, 4.6, 4.0, 4.4],
    'stock': [10, 0, 5, 8, 0, 3, 15, 20, 25, 7]
}
df = pd.DataFrame(data)
df
```

## **1.How many products are in each category?**

```python
df.groupby("category")["product"].count()
```

## **2.What is average price by brand?**

```python
df.groupby("brand")["price"].mean()
```

## **3.Which product has the highest rating?**

```python
df["rating"].max()
```

## **4.Filter products from Sony or Apple that are in stock?**

```python
df[df["brand"].isin(["Sony","Apple"]) & (df["stock"] > 0)]
```

## **5.Filter products NOT in Accessories with price above 10000?**

```python
df[~df["category"].isin(["Accessories"]) & (df["price"] > 10000)]
```
