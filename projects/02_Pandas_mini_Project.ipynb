```python
import pandas as pd

data = {
    'student': ['Aarav', 'Sneha', 'Ranjit', 'Priya', 'Karan', 'Anjali', 'Dev', 'Meera', 'Rohan', 'Tanvi'],
    'gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'city': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Nagpur', 'Mumbai', 'Delhi', 'Pune', 'Nagpur', 'Mumbai'],
    'marks': [88, 45, 72, 91, 55, 78, 60, 83, 47, 69],
    'attendance': [95, 60, 80, 98, 55, 88, 72, 91, 50, 75],
    'passed': [True, False, True, True, False, True, True, True, False, True]
}
df = pd.DataFrame(data)
df
```

## **1. How many students passed and failed?**

```python
df["passed"].value_counts()
```

## **What is the average marks of male vs female students?**

```python
df.groupby("gender")["marks"].mean()
```

## **3. Which city has the highest average attendance?**

```python
df.groupby("city")["attendance"].mean().idxmax()
```

## 4.**Filter students from Pune or Mumbai who passed**

```python
df[df["city"].isin(["Pune", "Mumbai"]) & (df["passed"] == True)]
```

## **5.Filter students whose name contains 'an' and marks above 60**

```python
df[df["student"].str.contains("an") & (df["marks"] > 60)]
```
