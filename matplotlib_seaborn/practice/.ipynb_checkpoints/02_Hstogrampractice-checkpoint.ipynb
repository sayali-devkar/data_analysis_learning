**Question:**

**This is the age data of subscribers of a cooking YouTube channel. Plot a histogram with bins=6 and tell me which age group watches the most!**

```python
import pandas as pd
import  matplotlib.pyplot as plt
```

```python
ages = [22, 25, 27, 19, 31, 35, 28, 22, 24, 30, 
        45, 42, 38, 50, 55, 48, 23, 26, 29, 33,
        21, 20, 19, 35, 40, 41, 44, 52, 58, 60]
```

```python
bins = 6
plt.hist(ages , bins = bins , edgecolor = "black")
plt.title('Age Distribution of Cooking Channel Viewers')
plt.xlabel('Age Group')
plt.ylabel('Number of Viewers')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
