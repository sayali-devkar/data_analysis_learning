```python
import matplotlib.pyplot as plt
import seaborn as sns
df = sns.load_dataset('tips')
df.head()
```

### **1. Plot a bar chart showing the total bill amount for each day (Sun, Sat, Fri, Thur)**

```python
grouped = df.groupby('day')['total_bill'].sum()
plt.bar(grouped.index, grouped.values , color = "skyblue")
plt.xlabel("Day")
plt.ylabel("Total_Bill")
plt.title("Which day we sale more?")
plt.show()
```

### **2. average tip by smokers vs non-smokers!**

```python
grouped = df.groupby('smoker')['tip'].mean()
plt.bar(grouped.index, grouped.values , color = "pink")
plt.xlabel("Category")
plt.ylabel("Average")
plt.title("average tip by smokers vs non-smokers")
plt.show()
```

### **3. horizontal bar chart of average total bill by party size!**

```python
grouped = df.groupby('size')['total_bill'].mean()
plt.barh(grouped.index, grouped.values , color = "#994B68")
plt.xlabel("Size")
plt.ylabel("Average")
plt.title("horizontal bar chart of average total bill by party size!")
plt.show()
```

### **4. grouped bar chart, average tip per day separated by gender**

```python
grouped = df.groupby(['day', 'sex'])['tip'].mean()

plt.xlabel("gender")
plt.ylabel("average tip per day")
plt.title("grouped bar chart, average tip per day separated by gender")
plt.show()
```

```python
sns.barplot(data=df, x='day', y='tip', hue='sex')
plt.show()
```

```python
grouped = df.groupby("day")["size"].count()
grouped = grouped.sort_values(ascending=False)
plt.bar(grouped.index, grouped.values, color="#86BF9C")
plt.xlabel("Day")
plt.ylabel("Customer Count")
plt.title("Which day is a rush day !")
plt.show()
```
