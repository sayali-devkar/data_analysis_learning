```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd

df = sns.load_dataset('tips')
df.head()
```

## **1. SCATTER PLOT**

```python
plt.scatter(df['total_bill'],df['tip'])
plt.xlabel('Total_Bill')
plt.ylabel('Tip')
plt.title('Does Bigger bill = more tip')
plt.show()
            
```

## **2. HISTOGRAM**

```python
plt.hist(df['total_bill'], bins=20, edgecolor='black')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.title('Distribution of Total Bill')
plt.show()
```

## **3.BAR GRAPH**

```python
plt.bar(df["day"],df["tip"])
plt.xlabel("Day")
plt.ylabel("Tip")
plt.title("How many people gave tip on the day ")
plt.show()
```

## **3.1 BAR GRAPH**

```python
grouped = df.groupby('day')['tip'].mean()
plt.bar(grouped.index , grouped.values , color = '#3498DB')
plt.xlabel('Day')
plt.ylabel('Average Tip')
plt.title("Average Tip by Day")
plt.show()
```

```python
sizes = df['smoker'].value_counts()
plt.pie(sizes.values, labels=sizes.index, autopct='%1.1f%%', colors=['#2ECC71','#E74C3C'])
plt.title('Smokers vs Non-Smokers')
plt.show()
```

```python
grouped = df.groupby('size')['total_bill'].mean()
plt.plot(grouped.index , grouped.values , color='#9B59B6', marker='o')
plt.xlabel('Table Size')
plt.ylabel('Average Total Bill')
plt.title('Does bigger table = bigger bill?')
plt.show()
```
